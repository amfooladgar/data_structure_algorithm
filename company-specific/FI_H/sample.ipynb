You are given an integer array coins representing coins of different denominations and an integer amount representing a total amount of money.

Return the fewest number of coins that you need to make up that amount. If that amount of money cannot be made up by any combination of the coins, return -1.

You may assume that you have an infinite number of each kind of coin.

 

Example 1:

Input: coins = [1,2,5], amount = 11
Output: 3
Explanation: 11 = 5 + 5 + 1
Example 2:

Input: coins = [2], amount = 3
Output: -1
Example 3:

Input: coins = [1], amount = 0
Output: 0
 

Constraints:

1 <= coins.length <= 12
1 <= coins[i] <= 231 - 1
0 <= amount <= 104

`````python
# using memoization: (note that this cannot be solved by greedy algorithm)
def coinChange(coins, amount):
    memo = {}

    def minCoin(val):
        if val == 0:
            return 0
        if val in memo:
            return memo[val]
        min_coin = float('inf')
        for coin in coins:
            if val - coin >=0:
                res = minCoin(val - coin)
                if res != float('inf'):
                    min_coin = min(min_coin, res + 1)
            memo[val] = min_coin
        return memo[val]

    return minCoin(amount) if minCoin(amount)!=float('inf') else -1

`````

`````python
coins = [1,2,5]
amount = 11
coinChange(coins, amount)
`````

`````python
coins = [2]
amount = 3
coinChange(coins, amount)
`````

`````python
# bottom-up approach
def coinChange(coins, amount):

    dp = [float('inf')]*(amount+1)
    dp[0] = 0

    for val in range(1, amount+1):
        for coin in coins:
            if val - coin >=0:
                dp[val] = min(dp[val], dp[val-coin]+1)
    
    return dp[amount] if dp[amount]!=float('inf') else -1

`````

`````python
coins = [1,2,5]
amount = 11
coinChange(coins, amount)
`````

`````python
coins = [2]
amount = 3
coinChange(coins, amount)
`````

## Another variant: 
If the coins are cash dominations {100, 50, 20, 10, 5, 1}, there is no need for dp, and can be done by greedy:

`````python
def coinChange(coins, amount):
    coins.sort(reverse=True)
    res = {}
    for coin in coins:
        count = amount // coin
        if count > 0:
            res[coin] = count

            amount %= coin
    return res
        
`````

You are given an integer array nums. You can choose exactly one index (0-indexed) and remove the element. Notice that the index of the elements may change after the removal.

For example, if nums = [6,1,7,4,1]:

Choosing to remove index 1 results in nums = [6,7,4,1].
Choosing to remove index 2 results in nums = [6,1,4,1].
Choosing to remove index 4 results in nums = [6,1,7,4].
An array is fair if the sum of the odd-indexed values equals the sum of the even-indexed values.

Return the number of indices that you could choose such that after the removal, nums is fair.

 

Example 1:

Input: nums = [2,1,6,4]
Output: 1
Explanation:
Remove index 0: [1,6,4] -> Even sum: 1 + 4 = 5. Odd sum: 6. Not fair.
Remove index 1: [2,6,4] -> Even sum: 2 + 4 = 6. Odd sum: 6. Fair.
Remove index 2: [2,1,4] -> Even sum: 2 + 4 = 6. Odd sum: 1. Not fair.
Remove index 3: [2,1,6] -> Even sum: 2 + 6 = 8. Odd sum: 1. Not fair.
There is 1 index that you can remove to make nums fair.
Example 2:

Input: nums = [1,1,1]
Output: 3
Explanation: You can remove any index and the remaining array is fair.
Example 3:

Input: nums = [1,2,3]
Output: 0
Explanation: You cannot make a fair array after removing any index.
 

Constraints:

1 <= nums.length <= 105
1 <= nums[i] <= 104

Hint 1
The parity of the indices after the removed element changes.
Hint 2
Calculate prefix sums for even and odd indices separately to calculate for each index in O(1).


`````python

def waysToMakeFair(nums):
    left = [0, 0]
    right = [sum(nums[::2]), sum(nums[1::2])]
    result = 0

    for i, num in enumerate(nums):
        parity = i % 2

        # Temporarily remove nums[i].
        right[parity] -= num

        # Elements on the right switch parity after removal.
        if left[0] + right[1] == left[1] + right[0]:
            result += 1

        # Move nums[i] to the left for the next iteration.
        left[parity] += num

    return result
`````

# Cohort Selection

Cancer researchers often want to use existing data to answer questions about cancer treatments. For example, physicians usually record lab results and treatments for patients in an electronic medical record, and this data can be aggregated to determine how effective different treatments are.

To answer these questions, researchers need to identify patients who have certain clinical characteristics in common. Such a collection of patients is called a **cohort**.

A cohort might consist of patients who:

- Were diagnosed with a particular illness.
- Were prescribed one or more specific drugs.

Your job is to build a tool that identifies which patients should be included in each cohort.

The tool receives:

1. A list of patient illnesses.
2. A list of drugs prescribed to patients.
3. A list of cohort criteria.

For each cohort, return all patients who satisfy both the illness requirement and all drug requirements.

---

```text
<patient identifier>, <illness>
```


```text
<patient identifier>, <drug name>
```

```text
patient1, drug1
```

A patient may have multiple drug records.

---

Each element in the `cohorts` array has the following format:

```text
<illness>, <drug name>, <drug name>, ...
```

- Has been diagnosed with melanoma.


If no patients match a cohort, return:

```text
[]
```

# Solution

## Intuition

We need to repeatedly answer two questions:

1. Which patients have a particular illness?
2. Which drugs has each patient received?

Instead of scanning all input records for every cohort, we can build two lookup tables:

```text
illness -> patients
patient -> drugs
```

Then, for each cohort, we only examine patients with the required illness and check whether they have all required drugs.

---

## Approach

Create a dictionary named `illness_to_patients`.

Each illness maps to a set of patients diagnosed with that illness:

```text
melanoma -> {patient1, patient3}
glioblastoma -> {patient2}
```

Create another dictionary named `patient_to_drugs`.

Each patient maps to a set of drugs prescribed to that patient:

```text
patient1 -> {drug1}
patient2 -> {drug1}
patient3 -> {drug1, drug2}
```

For every cohort:

1. Split the cohort string by commas.
2. Treat the first item as the required illness.
3. Treat the remaining items as the required drugs.
4. Find patients with the required illness.
5. Check whether every required drug is in each patient's drug set.
6. Sort the matching patients.
7. Format them as a bracketed string.

## Complexity

Let:

* `I` be the number of illness records.
* `D` be the number of drug records.
* `C` be the number of cohorts.
* `P` be the number of patients examined for a cohort.
* `R` be the number of required drugs in a cohort.

Building the lookup dictionaries takes:

```text
O(I + D)
```

Processing the cohorts takes approximately:

```text
O(C × P × R)
```

Sorting the matching patients for each cohort takes:

```text
O(K log K)
```

where `K` is the number of matching patients.

The space complexity is:

```text
O(I + D)
```

`````python
from collections import defaultdict

def solution(illnesses, drugs, cohorts):
    # illness -> patients diagnosed with that illness
    illness_to_patients = defaultdict(set)

    # patient -> drugs prescribed to that patient
    patient_to_drugs = defaultdict(set)

    # Process diagnoses.
    for record in illnesses:
        patient, illness = record.split(",")

        patient = patient.strip()
        illness = illness.strip()

        illness_to_patients[illness].add(patient)

    # Process prescriptions.
    for record in drugs:
        patient, drug = record.split(",")

        patient = patient.strip()
        drug = drug.strip()

        patient_to_drugs[patient].add(drug)

    result = []

    # Process each cohort in its original order.
    for cohort in cohorts:
        parts = cohort.split(",")

        required_illness = parts[0].strip()
        required_drugs = set()

        for i in range(1, len(parts)):
            required_drugs.add(parts[i].strip())

        matching_patients = []

        # Only examine patients with the required illness.
        for patient in illness_to_patients[required_illness]:
            patient_drugs = patient_to_drugs[patient]

            # The patient must have every required drug.
            if required_drugs.issubset(patient_drugs):
                matching_patients.append(patient)

        matching_patients.sort()

        formatted_result = "[" + ", ".join(matching_patients) + "]"
        result.append(formatted_result)

    return result
`````

`````python
## Simpler version

from collections import defaultdict


def solution(illnesses, drugs, cohorts):
    illness_to_patients = defaultdict(set)
    patient_to_drugs = defaultdict(set)

    for record in illnesses:
        patient, illness = map(str.strip, record.split(","))
        illness_to_patients[illness].add(patient)

    for record in drugs:
        patient, drug = map(str.strip, record.split(","))
        patient_to_drugs[patient].add(drug)

    answer = []

    for cohort in cohorts:
        criteria = [item.strip() for item in cohort.split(",")]

        illness = criteria[0]
        required_drugs = set(criteria[1:])

        patients = []

        for patient in illness_to_patients[illness]:
            if required_drugs <= patient_to_drugs[patient]:
                patients.append(patient)

        patients.sort()
        answer.append("[" + ", ".join(patients) + "]")

    return answer
`````
