You are given an integer array coins representing coins of different denominations and an integer amount representing a total amount of money.

Return the fewest number of coins that you need to make up that amount. If that amount of money cannot be made up by any combination of the coins, return -1.

You may assume that you have an infinite number of each kind of coin.

 

Example 1:

Input: coins = [1,2,5], amount = 11
Output: 3
Explanation: 11 = 5 + 5 + 1
Example 2:

Input: coins = [2], amount = 3
Output: -1
Example 3:

Input: coins = [1], amount = 0
Output: 0
 

Constraints:

1 <= coins.length <= 12
1 <= coins[i] <= 231 - 1
0 <= amount <= 104

```python
# using memoization: (note that this cannot be solved by greedy algorithm)
def coinChange(coins, amount):
    memo = {}

    def minCoin(val):
        if val == 0:
            return 0
        if val in memo:
            return memo[val]
        min_coin = float('inf')
        for coin in coins:
            if val - coin >=0:
                res = minCoin(val - coin)
                if res != float('inf'):
                    min_coin = min(min_coin, res + 1)
            memo[val] = min_coin
        return memo[val]

    return minCoin(amount) if minCoin(amount)!=float('inf') else -1

```

```python
coins = [1,2,5]
amount = 11
coinChange(coins, amount)
```

```python
coins = [2]
amount = 3
coinChange(coins, amount)
```

```python
# bottom-up approach
def coinChange(coins, amount):

    dp = [float('inf')]*(amount+1)
    dp[0] = 0

    for val in range(1, amount+1):
        for coin in coins:
            if val - coin >=0:
                dp[val] = min(dp[val], dp[val-coin]+1)
    
    return dp[amount] if dp[amount]!=float('inf') else -1

```

```python
coins = [1,2,5]
amount = 11
coinChange(coins, amount)
```

```python
coins = [2]
amount = 3
coinChange(coins, amount)
```

## Another variant: 
If the coins are cash dominations {100, 50, 20, 10, 5, 1}, there is no need for dp, and can be done by greedy:

```python
def coinChange(coins, amount):
    coins.sort(reverse=True)
    res = {}
    for coin in coins:
        count = amount // coin
        if count > 0:
            res[coin] = count

            amount %= coin
    return res
        
```
