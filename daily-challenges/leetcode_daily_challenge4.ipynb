A prefix nums[0..i] is sequential if, for all 1 <= j <= i, nums[j] = nums[j - 1] + 1. In particular, the prefix consisting only of nums[0] is sequential.

Return the smallest integer x missing from nums such that x is greater than or equal to the sum of the longest sequential prefix.

 

Example 1:

Input: nums = [1,2,3,2,5]
Output: 6
Explanation: The longest sequential prefix of nums is [1,2,3] with a sum of 6. 6 is not in the array, therefore 6 is the smallest missing integer greater than or equal to the sum of the longest sequential prefix.

Output: 15
Explanation: The longest sequential prefix of nums is [3,4,5] with a sum of 12. 12, 13, and 14 belong to the array while 15 does not. Therefore 15 is the smallest missing integer greater than or equal to the sum of the longest sequential prefix.
 

Constraints:

1 <= nums.length <= 50
1 <= nums[i] <= 50

Hint 1
To find the longest sequential prefix, iterate from left to right. For a fixed i, if nums[i] != nums[i - 1] + 1 then the longest sequential prefix ends at i - 1.

````python
# Approach 1: Simulation + Prefix Sum: O(n) time and space
def missingInteger(nums):
    n = len(nums)
    total = nums[0]
    for i in range(1,n):
        if nums[i-1]+1==nums[i]:
            total += nums[i]
        else:
            break
    nums_set = set(nums)
    while total in nums_set:
        total +=1
    return total
````

````python
# Approach 2: Simulation + Series Sum Formula
def missingInteger(nums):
    nums_set = set(nums)
    prev = nums[0]
    prefix_len = 1
    for num in nums[1:]:
        if prev + 1 == num:
            prefix_len += 1
            prev = num
        else:
            break
    total = (nums[prefix_len - 1] + nums[0]) * prefix_len // 2
    while total in nums_set:
        total+=1
    return total
````

````python
# Approach 1: Counting and Sliding Window: O(n) time and O(n) space (worst case)
from collections import Counter
def maxSubarrayLength(nums, k):
    n = len(nums)
    freq = Counter()

    ans, start = 0, 0
    for end in range(n):
        freq[nums[end]] += 1
        while freq[nums[end]] > k:
            freq[nums[start]] -= 1
            start+=1
        ans = max(ans, end - start+1)
    return ans
````

````python
# Approach 2: Counting and Sliding Window without Nested Loops

from collections import Counter
def maxSubarrayLength(nums, k):
    n = len(nums)
    freq = Counter()
    start = 0
    freq_more_k = 0
    for end in range(n):
        freq[nums[end]]+=1
        if freq[nums[end]] == k + 1:
            freq_more_k += 1
        if freq_more_k:
            freq[nums[start]] -= 1
            if freq[nums[start]] == k: # this guarantees that the size of our current window is equal to the largest "good" subarray encountered so far.
                freq_more_k -=1
            start += 1
    return n - start 
````

# Intuition

We use a sliding window to find the longest subarray where each number appears at most `k` times.

The window grows when it is valid. When a number appears more than `k` times, we move `start` forward once to prevent the window size from increasing.

Unlike the standard sliding-window solution, we use `if` instead of `while`. Therefore, the current window can remain temporarily invalid. This is safe because an invalid window never grows:

* Valid iteration: only `end` moves, so the window grows.
* Invalid iteration: both `end` and `start` move, so the window size stays unchanged.

Thus, the final window size `n - start` is the maximum valid length encountered.

# Approach

1. Use `freq` to count each number in the current window.

2. Use `violations` to count how many values occur more than `k` times.

3. For every `nums[end]`:

   * Add it to the window.
   * If its frequency becomes exactly `k + 1`, add one violation.
   * If a violation exists, remove `nums[start]` and move `start` once.
   * If the removed number’s frequency becomes `k`, that violation is resolved.

4. Return `n - start`, which represents the largest valid window size encountered.

```python
from collections import Counter


class Solution:
    def maxSubarrayLength(self, nums: List[int], k: int) -> int:
        n = len(nums)
        freq = Counter()
        start = 0
        violations = 0

        for end in range(n):
            freq[nums[end]] += 1

            # nums[end] has just exceeded the allowed frequency.
            if freq[nums[end]] == k + 1:
                violations += 1

            # Move start once so an invalid window cannot grow.
            if violations:
                freq[nums[start]] -= 1

                # This number is no longer violating the limit.
                if freq[nums[start]] == k:
                    violations -= 1

                start += 1

        return n - start
```

# Complexity

* Time complexity: (O(n))

  Each element is added once and removed at most once.

* Space complexity: (O(u))

  Here, (u) is the number of unique values in `nums`.

# Step-by-step solution derivation

## 1. Why a simple scan is too slow

After each update, we could scan the full string and find the longest repeating run.

* One scan takes (O(n)).
* There can be (10^5) updates.
* Total time can reach (O(nk)), or approximately (10^{10}) operations.

We need to update one position without recalculating the full string.

## 2. Use a segment tree

A segment tree divides the string into smaller intervals.

After changing one character:

* Update its leaf.
* Recalculate only its ancestors.
* Read the answer from the root.

This gives (O(\log n)) time for each query.

## 3. What must each node remember?

For its substring, each node stores:

| Field        | Meaning                                     |
| ------------ | ------------------------------------------- |
| `length`     | Number of characters in the segment         |
| `left_char`  | First character                             |
| `right_char` | Last character                              |
| `prefix`     | Longest uniform run starting from the left  |
| `suffix`     | Longest uniform run ending at the right     |
| `best`       | Longest uniform run anywhere in the segment |

For a leaf containing one character:

```text
length = prefix = suffix = best = 1
left_char = right_char = character
```

## 4. Merge two children

Suppose an interval has children `L` and `R`.

The combined run can cross their boundary only when:

```python
L.right_char == R.left_char
```

Start with values that do not cross the boundary:

```python
prefix = L.prefix
suffix = R.suffix
best = max(L.best, R.best)
```

If the boundary characters match, the crossing run is:

```python
L.suffix + R.prefix
```

Therefore:

```python
best = max(best, L.suffix + R.prefix)
```

The prefix can extend into `R` only when the complete left segment is uniform:

```python
if L.prefix == L.length:
    prefix = L.length + R.prefix
```

Similarly, the suffix can extend into `L` only when the complete right segment is uniform:

```python
if R.suffix == R.length:
    suffix = R.length + L.suffix
```

## 5. Process an update

For every query:

1. Change the corresponding leaf.
2. Recalculate the nodes on the path to the root.
3. Append `tree[1].best` to the result.

The root represents the complete string, so its `best` value is the required answer.

---

# Intuition

A character update can affect only the repeating runs that pass through that position. A segment tree lets us update that position and recompute only the affected intervals.

For each interval, we remember its best internal run and the uniform runs touching its two boundaries. When two intervals are merged, their boundary runs can connect only when the touching characters are equal.

# Approach

Build a segment tree over the characters of `s`.

Each node stores:

* The segment length.
* Its leftmost and rightmost characters.
* The longest uniform prefix.
* The longest uniform suffix.
* The longest uniform substring anywhere in the segment.

To merge nodes `L` and `R`:

1. Copy the prefix from `L` and the suffix from `R`.
2. Set the initial best result to `max(L.best, R.best)`.
3. If `L.right_char == R.left_char`:

   * Consider `L.suffix + R.prefix` as a crossing run.
   * Extend the prefix if all of `L` is uniform.
   * Extend the suffix if all of `R` is uniform.

For each query, update one leaf and recompute its ancestors. The root node’s `best` value is the answer after that query.

# Complexity

* Time complexity:

  Building the tree takes (O(n)).

  Each point update takes (O(\log n)), so all queries take:

  [
  O(n + k\log n)
  ]

* Space complexity:

  [
  O(n)
  ]

  The segment tree contains (O(n)) nodes.


Memory rule: **Two runs connect only when their boundary characters match. The prefix or suffix extends further only when the corresponding child is completely uniform.**


````python
from typing import List


class Node:
    __slots__ = (
        "length",
        "left_char",
        "right_char",
        "prefix",
        "suffix",
        "best",
    )

    def __init__(
        self,
        length=0,
        left_char="",
        right_char="",
        prefix=0,
        suffix=0,
        best=0,
    ):
        self.length = length
        self.left_char = left_char
        self.right_char = right_char
        self.prefix = prefix
        self.suffix = suffix
        self.best = best


class Solution:
    def longestRepeating(
        self,
        s: str,
        queryCharacters: str,
        queryIndices: List[int],
    ) -> List[int]:

        n = len(s)
        characters = list(s)
        tree = [Node() for _ in range(4 * n)]

        def make_leaf(char: str) -> Node:
            return Node(
                length=1,
                left_char=char,
                right_char=char,
                prefix=1,
                suffix=1,
                best=1,
            )

        def merge(left: Node, right: Node) -> Node:
            node = Node()

            node.length = left.length + right.length
            node.left_char = left.left_char
            node.right_char = right.right_char

            # Runs that do not cross the boundary.
            node.prefix = left.prefix
            node.suffix = right.suffix
            node.best = max(left.best, right.best)

            # Boundary runs can connect only when the characters match.
            if left.right_char == right.left_char:
                crossing_run = left.suffix + right.prefix
                node.best = max(node.best, crossing_run)

                # The prefix reaches the right child only when
                # the complete left child is uniform.
                if left.prefix == left.length:
                    node.prefix = left.length + right.prefix

                # The suffix reaches the left child only when
                # the complete right child is uniform.
                if right.suffix == right.length:
                    node.suffix = right.length + left.suffix

            return node

        def build(tree_index: int, segment_left: int, segment_right: int) -> None:
            if segment_left == segment_right:
                tree[tree_index] = make_leaf(characters[segment_left])
                return

            middle = (segment_left + segment_right) // 2

            build(tree_index * 2, segment_left, middle)
            build(tree_index * 2 + 1, middle + 1, segment_right)

            tree[tree_index] = merge(
                tree[tree_index * 2],
                tree[tree_index * 2 + 1],
            )

        def update(
            tree_index: int,
            segment_left: int,
            segment_right: int,
            position: int,
            new_character: str,
        ) -> None:
            if segment_left == segment_right:
                tree[tree_index] = make_leaf(new_character)
                return

            middle = (segment_left + segment_right) // 2

            if position <= middle:
                update(
                    tree_index * 2,
                    segment_left,
                    middle,
                    position,
                    new_character,
                )
            else:
                update(
                    tree_index * 2 + 1,
                    middle + 1,
                    segment_right,
                    position,
                    new_character,
                )

            tree[tree_index] = merge(
                tree[tree_index * 2],
                tree[tree_index * 2 + 1],
            )

        build(1, 0, n - 1)

        answers = []

        for position, new_character in zip(
            queryIndices,
            queryCharacters,
        ):
            characters[position] = new_character
            update(1, 0, n - 1, position, new_character)
            answers.append(tree[1].best)

        return answers
````

````python
# Leetcode solution:
class Solution:
    def longestRepeating(
        self, s: str, queryCharacters: str, queryIndices: List[int]
    ) -> List[int]:
        n = len(s)
        pre = [0] * (4 * n)
        suf = [0] * (4 * n)
        maxLen = [0] * (4 * n)
        leftChar = [""] * (4 * n)
        rightChar = [""] * (4 * n)

        def build(u: int, l: int, r: int) -> None:
            if l == r:
                pre[u] = 1
                suf[u] = 1
                maxLen[u] = 1
                leftChar[u] = s[l]
                rightChar[u] = s[l]
                return
            mid = (l + r) >> 1
            build(u << 1, l, mid)
            build(u << 1 | 1, mid + 1, r)
            pushUp(u, l, r)

        def pushUp(u: int, l: int, r: int) -> None:
            mid = (l + r) >> 1
            leftLen = mid - l + 1
            rightLen = r - mid
            left = u << 1
            right = u << 1 | 1
            leftChar[u] = leftChar[left]
            rightChar[u] = rightChar[right]
            pre[u] = pre[left]
            if pre[left] == leftLen and rightChar[left] == leftChar[right]:
                pre[u] = pre[left] + pre[right]
            suf[u] = suf[right]
            if suf[right] == rightLen and rightChar[left] == leftChar[right]:
                suf[u] = suf[right] + suf[left]
            maxLen[u] = max(maxLen[left], maxLen[right])
            if rightChar[left] == leftChar[right]:
                maxLen[u] = max(maxLen[u], suf[left] + pre[right])

        def update(u: int, l: int, r: int, pos: int, ch: str) -> None:
            if l == r:
                leftChar[u] = ch
                rightChar[u] = ch
                return
            mid = (l + r) >> 1
            if pos <= mid:
                update(u << 1, l, mid, pos, ch)
            else:
                update(u << 1 | 1, mid + 1, r, pos, ch)
            pushUp(u, l, r)

        build(1, 0, n - 1)
        k = len(queryIndices)
        ans = []
        for i in range(k):
            update(1, 0, n - 1, queryIndices[i], queryCharacters[i])
            ans.append(maxLen[1])
        return ans
````

````python
# Brute force/enumeration: O(n^2) time and O(1) space
def maximumLengthSubstring(s):
    n = len(s)
    ans = 0
    for l in range(n):
        count = [0]*26
        for r in range(l, n):
            ch = ord(s[r]) - ord('a')
            count[ch] += 1
            if count[ch]>2:
                break
            ans = max(ans, r - l + 1)
    return ans
````

````python
from collections import defaultdict

def maximumLengthSubstring(s: str) -> int:
    count = defaultdict(int)
    left = 0
    ans = 0

    for right, char in enumerate(s):
        count[char] += 1

        # Shrink until the current character appears at most twice.
        while count[char] > 2:
            count[s[left]] -= 1
            left += 1

        ans = max(ans, right - left + 1)

    return ans
````

````python
# Approach: Classification Discussion
def longestSubsequence(nums):
    totalXor = 0
    n = len(nums)
    allZero = True
    for num in nums:
        totalXor ^= num
        if num > 0:
            allZero = False

    if totalXor > 0:
        return n
    return n - 1 if allZero == False else 0
````

Alice and Bob continue their games with stones. There is a row of n stones, and each stone has an associated value. You are given an integer array stones, where stones[i] is the value of the ith stone.

The player who removes a stone loses if the sum of the values of all removed stones is divisible by 3. Bob will win automatically if there are no remaining stones (even if it is Alice's turn).

Assuming both players play optimally, return true if Alice wins and false if Bob wins.

 

Example 1:

Input: stones = [2,1]
Output: true
Explanation: The game will be played as follows:
- Turn 1: Alice can remove either stone.
- Turn 2: Bob removes the remaining stone. 
The sum of the removed stones is 1 + 2 = 3 and is divisible by 3. Therefore, Bob loses and Alice wins the game.
Example 2:

- Turn 1: Alice can remove the second stone with value 1. Sum of removed stones = 1.

- Turn 3: Alices removes the fourth stone with value 4. Sum of removed stones = 1 + 3 + 4 = 8.

1 <= stones.length <= 105
1 <= stones[i] <= 104

Hint 1
There are limited outcomes given the current sum and the stones remaining.
Hint 2
Can we greedily simulate starting with taking a stone with remainder 1 or 2 divided by 3?

````python
from typing import List


class Solution:
    def stoneGameIX(self, stones: List[int]) -> bool:
        count = [0, 0, 0]

        for stone in stones:
            count[stone % 3] += 1

        c0, c1, c2 = count

        if c0 % 2 == 0:
            return c1 > 0 and c2 > 0

        return abs(c1 - c2) > 2
````

### Straight solution
![3471-find-the-largest-almost-missing-integer-2026-08-18.png](../assets/images/3471-find-the-largest-almost-missing-integer-2026-08-18.png)

````python
# sliding window: complexity O(nk) time and O(n) space
from collections import defaultdict
def largestInteger(nums, k):
    n = len(nums)
    seen = defaultdict(int)

    for i in range(n - k + 1):
        # Count each number at most once per subarray
        unique_nums = set(nums[i:i + k])

        for num in unique_nums:
            seen[num] += 1

    ans = -1

    for num, count in seen.items():
        if count == 1:
            ans = max(ans, num)

    return ans
````

````python
# Greedy-Bitwise Operations: Complexity: Let m be the length of the array reservedSeats. O(m) time and O(m) space
from collections import defaultdict
from typing import List


class Solution:
    def maxNumberOfFamilies(
        self,
        n: int,
        reservedSeats: List[List[int]]
    ) -> int:
        reserved_by_row = defaultdict(int)

        # Record only seats that can block a four-seat group.
        for row, seat in reservedSeats:
            if 2 <= seat <= 9:
                reserved_by_row[row] |= 1 << seat

        left_mask = sum(1 << seat for seat in range(2, 6))    # 2–5
        middle_mask = sum(1 << seat for seat in range(4, 8))  # 4–7
        right_mask = sum(1 << seat for seat in range(6, 10))  # 6–9

        # Every unaffected row can hold two groups.
        answer = 2 * (n - len(reserved_by_row))

        for reserved_mask in reserved_by_row.values():
            left_free = (reserved_mask & left_mask) == 0
            middle_free = (reserved_mask & middle_mask) == 0
            right_free = (reserved_mask & right_mask) == 0

            if left_free and right_free:
                answer += 2
            elif left_free or middle_free or right_free:
                answer += 1

        return answer
````

Constraints:

1 <= coins.length <= 15
1 <= coins[i] <= 25
1 <= k <= 2 * 109
coins contains pairwise distinct integers.

Hint 1
Binary search the answer x.
Hint 2
Use the inclusion-exclusion principle to count the number of distinct amounts that can be made up to x.

![3116-kth-smallest-amount-with-single-denomination-combination-2026-08-21](../assets/images/3116-kth-smallest-amount-with-single-denomination-combination-2026-08-21.png)

The inclusion-exclusion principle counts how many distinct numbers up to `x` are divisible by at least one coin.

For each coin, define a set:

$$
A_i = {m \le x \mid m \text{ is divisible by } coins[i]}
$$

We need the size of their union:

$$
|A_1 \cup A_2 \cup \cdots \cup A_n|
$$

We cannot only add their sizes because some numbers occur in multiple sets.

For example, with `coins = [2, 3]` and `x = 10`:

* Multiples of 2: `{2, 4, 6, 8, 10}` → 5
* Multiples of 3: `{3, 6, 9}` → 3

Adding gives `5 + 3 = 8`, but `6` was counted twice. Therefore, subtract the intersection:

$
5 + 3 - 1 = 7
$

The intersection contains numbers divisible by both coins. These numbers are multiples of their LCM:

$$
|A_i \cap A_j| = \left\lfloor\frac{x}{\operatorname{lcm}(coins[i], coins[j])}\right\rfloor
$$

The general formula is:

$$
\begin{aligned}
|A_1 \cup \cdots \cup A_n|
={}&\sum |A_i| \
&-\sum |A_i \cap A_j| \
&+\sum |A_i \cap A_j \cap A_k| \
&-\cdots
\end{aligned}
$$

So:

* Subsets with an odd number of coins are added.
* Subsets with an even number of coins are subtracted.

### How the code implements it

```python
for mask in range(1, 1 << n):
```

Each nonempty bitmask represents one subset of coins.

For example, with three coins:

| Mask  | Selected coins  | Operation |
| ----- | --------------- | --------- |
| `001` | coin 0          | Add       |
| `010` | coin 1          | Add       |
| `011` | coins 0 and 1   | Subtract  |
| `101` | coins 0 and 2   | Subtract  |
| `111` | all three coins | Add       |

For each subset, the code calculates its LCM:

```python
LCM = LCM * coins[i] // gcd(LCM, coins[i])
```

Then:

```python
term = x // LCM
```

This is the number of values up to `x` divisible by all coins in that subset.

Finally:

```python
total += term if bits % 2 else -term
```

* Odd subset size → add
* Even subset size → subtract

### Complete example

For `coins = [2, 3, 5]` and `x = 10`:

$$
\begin{aligned}
count(10)
={}&\lfloor10/2\rfloor+\lfloor10/3\rfloor+\lfloor10/5\rfloor\
&-\lfloor10/6\rfloor-\lfloor10/10\rfloor-\lfloor10/15\rfloor\
&+\lfloor10/30\rfloor\
={}&5+3+2-1-1-0+0\
={}&8
\end{aligned}
$$

The eight distinct valid numbers are:

```text
2, 3, 4, 5, 6, 8, 9, 10
```

Inclusion-exclusion is necessary because the problem counts distinct amounts, not the number of ways an amount can be produced. Without it, values such as `6`, which are divisible by both `2` and `3`, would be counted more than once.

The binary search then finds the smallest `x` for which:

```python
count_le(x) >= k
```

That `x` is the kth smallest valid amount.


````python
# Approach 1: Binary Answer + Inclusion-Exclusion Principle: comlexity: O(2^n ×(nlog(max{coins})+log(k×min{coins}))): 
# During the preprocessing stage, calculating the least common multiple of each subset takes O(nlog(maxcoins)) time per subset, 
# and there are O(2^n) subsets. During the binary search, we perform O(log(k×mincoins)) iterations, and each iteration 
# takes O(2^n) time to calculate count[x].

from math import lcm, gcd
def findKthSmallest(coins, k):
    n = len(coins)
    lo, hi = 0, min(coins)*k

    def count_le(x):
        total = 0
        for mask in range(1,1<<n):
            LCM, bits = 1, 0
            for i in range(n):
                if mask & (1<<i):
                    bits += 1
                    LCM = LCM * coins[i] // gcd(LCM, coins[i])
                    if LCM>x:
                        break
            else:
                term  = x // LCM
                total += term if bits % 2 else -term
        return total


    while lo<hi:
        mid = (lo+hi)//2
        if count_le(mid)<k:
            lo = mid + 1
        else:
            hi = mid
    return lo
    
````

````python
# Approach 2: Optimized Based on Approach 1
class Solution:
    def findKthSmallest(self, coins: List[int], k: int) -> int:
        coins.sort()
        new_coins = []
        for x in coins:
            if all(x % y for y in new_coins):
                new_coins.append(x)
        coins = new_coins

        n = len(coins)
        m = 1 << n
        lcm = [1] * m

        left = k
        right = coins[0] * k + 1

        for mask in range(1, m):
            pre_mask = mask & (mask - 1)
            i = (mask & -mask).bit_length() - 1

            tmp = lcm[pre_mask] // gcd(lcm[pre_mask], coins[i])
            if tmp <= right // coins[i]:
                lcm[mask] = tmp * coins[i]
            else:
                lcm[mask] = right + 1

        def get(x: int) -> int:
            count = 0
            for mask in range(1, m):
                if lcm[mask] > x:
                    continue
                if mask.bit_count() & 1:
                    count += x // lcm[mask]
                else:
                    count -= x // lcm[mask]

            return count

        while left < right:
            mid = (left + right) // 2
            if get(mid) >= k:
                right = mid
            else:
                left = mid + 1

        return left
````

````python
def stoneGameVIII(stones):
    n = len(stones)

    # Prefix sums
    prefix = [0] * n
    prefix[0] = stones[0]

    for i in range(1, n):
        prefix[i] = prefix[i - 1] + stones[i]

    # If Alice takes all stones, her advantage is the total sum.
    best = prefix[-1]

    # First move must take at least two stones, so stop at index 1.
    for i in range(n - 2, 0, -1):
        best = max(
            best,              # Skip this endpoint
            prefix[i] - best   # Take this prefix, then opponent plays
        )

    return best
````

![3720-lexicographically-smallest-permutation-greater-than-target-2026-08-27](../assets/images/3720-lexicographically-smallest-permutation-greater-than-target-2026-08-27.png)

````python
from collections import Counter

def lexGreaterPermutation(s: str, target: str) -> str:
    n = len(s)
    counts = Counter(s)
    prefix = []
    i = 0

    # Match as much of target as possible.
    while i < n and counts[target[i]] > 0:
        counts[target[i]] -= 1
        prefix.append(target[i])
        i += 1

    while True:
        # Try to make position i strictly greater.
        if i < n:
            for code in range(ord(target[i]) + 1, ord('z') + 1):
                ch = chr(code)

                if counts[ch] > 0:
                    counts[ch] -= 1

                    suffix = "".join(
                        chr(c) * counts[chr(c)]
                        for c in range(ord('a'), ord('z') + 1)
                    )

                    return "".join(prefix) + ch + suffix

        # No earlier matched position remains.
        if not prefix:
            return ""

        # Undo the most recent exact match and try to increase it.
        i -= 1
        restored = prefix.pop()
        counts[restored] += 1
````

Return the lexicographically smallest string that is both a palindromic permutation of s and strictly greater than target. If no such permutation exists, return an empty string.

"aca" is strictly greater than target. Therefore, the answer is "aca".
 

Constraints:

1 <= n == s.length == target.length <= 300
s and target consist of only lowercase English letters.

Hint 1
A palindromic permutation exists only if at most one character has an odd count (for odd-length strings) or all counts are even (for even-length strings).
Hint 2
Focus on constructing the first half of the palindrome. The second half is determined by mirroring.
Hint 3
To be lexicographically greater than target, the first half must be greater than or equal to target's first half, with careful handling of the middle character for odd-length strings.
Hint 4
Use a backtracking approach or greedy selection for each position in the first half, trying the smallest available character that can still produce a valid palindrome.
Hint 5
After building the first half, mirror it (and add the middle character if needed) to form the full palindrome and verify it is strictly greater than target.

![3734-lexicographically-smallest-palindromic-permutation-greater-than-target-2026-08-28](../assets/images/3734-lexicographically-smallest-palindromic-permutation-greater-than-target-2026-08-28.png)

````python
from collections import Counter


class Solution:
    def lexPalindromicPermutation(self, s: str, target: str) -> str:
        n = len(s)

        # Count the characters in s.
        freq = [0] * 26
        for ch in s:
            freq[ord(ch) - ord("a")] += 1

        # Check whether a palindromic permutation exists.
        odd = [i for i in range(26) if freq[i] % 2 == 1]

        if n % 2 == 0 and odd:
            return ""

        if n % 2 == 1 and len(odd) != 1:
            return ""

        middle = chr(odd[0] + ord("a")) if n % 2 else ""

        half = n // 2

        # Available characters for the left half.
        counts = [count // 2 for count in freq]

        # Rightmost position where target's prefix can be increased.
        last_position = -1
        last_character = -1
        exact_prefix = True

        for i in range(half):
            target_char = ord(target[i]) - ord("a")

            # Find the smallest available character greater than target[i].
            candidate = next(
                (
                    char
                    for char in range(target_char + 1, 26)
                    if counts[char] > 0
                ),
                -1,
            )

            if candidate != -1:
                last_position = i
                last_character = candidate

            # Continue matching target's left half.
            if counts[target_char] > 0:
                counts[target_char] -= 1
            else:
                exact_prefix = False
                break

        # Check the palindrome whose left half equals target's left half.
        if exact_prefix:
            left = target[:half]
            right = left[::-1]

            if n % 2 == 0:
                is_greater = right > target[half:]
            else:
                is_greater = (
                    middle > target[half]
                    or (
                        middle == target[half]
                        and right > target[half + 1:]
                    )
                )

            if is_greater:
                return left + middle + right

        # The left half cannot be increased.
        if last_position == -1:
            return ""

        # Rebuild the character counts before last_position.
        remaining = [count // 2 for count in freq]

        for i in range(last_position):
            char = ord(target[i]) - ord("a")
            remaining[char] -= 1

        # Use the smallest character that makes the prefix greater.
        remaining[last_character] -= 1

        # Fill the remaining left half in ascending order.
        tail = "".join(
            chr(char + ord("a")) * remaining[char]
            for char in range(26)
        )

        left = (
            target[:last_position]
            + chr(last_character + ord("a"))
            + tail
        )

        return left + middle + left[::-1]
````

````python
class Solution:
    def lexPalindromicPermutation(self, s: str, target: str) -> str:
        n = len(s)
        # Special case: length of 1
        if n == 1:
            return s if s > target else ""

        # Count the frequency of each character
        cnt = [0] * 26
        for c in s:
            cnt[ord(c) - ord("a")] += 1

        # Check if it can form a palindrome and record the characters with odd occurrences
        odd_char = ""
        for i in range(26):
            if cnt[i] % 2 == 1:
                # More than one character appears an odd number of times, cannot form a palindrome
                if odd_char != "":
                    return ""
                odd_char = chr(ord("a") + i)
            cnt[
                i
            ] //= 2  # It takes only half the characters to construct the left half

        prefix = []

        def check(c):
            left = prefix.copy()
            left.append(c)
            for i in range(25, -1, -1):
                left.extend([chr(ord("a") + i)] * cnt[i])

            palindrome = left + [odd_char] + left[::-1]

            return "".join(palindrome) > target

        # Construct the left part of each digit greedily
        for i in range(n // 2):
            found = False
            # Try to place the smallest character in lexicographical order
            for j in range(26):
                if cnt[j] == 0:
                    continue

                cnt[j] -= 1
                if check(chr(ord("a") + j)):
                    # If the constructed palindrome is greater than target, choose the character
                    prefix.append(chr(ord("a") + j))
                    found = True
                    break
                else:
                    cnt[j] += 1  # Not meeting the conditions, reset the counter
            if not found:
                return ""  # Cannot construct a palindrome larger than target

            if prefix[i] > target[i]:  # prefix is already greater than target
                left = prefix[:]
                for j in range(26):
                    left.extend([chr(ord("a") + j)] * cnt[j])
                palindrome = left + [odd_char] + left[::-1]
                return "".join(palindrome)

        # Construct the final palindrome string
        ans = prefix + [odd_char] + prefix[::-1]
        return "".join(ans)
````

An array a is lexicographically smaller than an array b if in the first position where a and b differ, array a has an element that is less than the corresponding element in b. For example, the array [2,10,3] is lexicographically smaller than the array [10,2,3] because they differ at index 0 and 2 < 10.

Explanation: [1,7,28,19,10] is the lexicographically smallest array we can obtain because we cannot apply the operation on any two indices.
 

Constraints:

1 <= nums.length <= 105
1 <= nums[i] <= 109
1 <= limit <= 109

Hint 1
Construct a virtual graph where all elements in nums are nodes and the pairs satisfying the condition have an edge between them.
Hint 2
Instead of constructing all edges, we only care about the connected components.
Hint 3
Can we use DSU?
Hint 4
Sort nums. Now we just need to consider if the consecutive elements have an edge to check if they belong to the same connected component. Hence, all connected components become a list of position-consecutive elements after sorting.
Hint 5
For each index of nums from 0 to nums.length - 1 we can change it to the current minimum value we have in its connected component and remove that value from the connected component.

````python
# DSU Solution: Complexity O(nlogn) time and O(n) space
from collections import defaultdict
from typing import List


class DSU:
    def __init__(self, n: int):
        self.parent = list(range(n))
        self.size = [1] * n

    def find(self, x: int) -> int:
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def union(self, a: int, b: int) -> None:
        root_a = self.find(a)
        root_b = self.find(b)

        if root_a == root_b:
            return

        if self.size[root_a] < self.size[root_b]:
            root_a, root_b = root_b, root_a

        self.parent[root_b] = root_a
        self.size[root_a] += self.size[root_b]


class Solution:
    def lexicographicallySmallestArray(
        self,
        nums: List[int],
        limit: int
    ) -> List[int]:

        n = len(nums)
        dsu = DSU(n)

        # Sort indices according to their values.
        sorted_indices = sorted(range(n), key=lambda i: nums[i])

        # Only consecutive sorted values must be checked.
        for k in range(1, n):
            previous_index = sorted_indices[k - 1]
            current_index = sorted_indices[k]

            if nums[current_index] - nums[previous_index] <= limit:
                dsu.union(previous_index, current_index)

        # Collect the indices in each connected component.
        components = defaultdict(list)

        for index in range(n):
            root = dsu.find(index)
            components[root].append(index)

        answer = nums[:]

        # Assign the smallest values to the smallest indices.
        for indices in components.values():
            values = sorted(nums[index] for index in indices)
            indices.sort()

            for index, value in zip(indices, values):
                answer[index] = value

        return answer
````

![2948-make-lexicographically-smallest-array-by-swapping-elements-2026-08-29](../assets/images/2948-make-lexicographically-smallest-array-by-swapping-elements-2026-08-29.png)

````python
from typing import List
def lexicographicallySmallestArray(nums, limit):
    n = len(nums)

    # Indices sorted by their corresponding values.
    order = sorted(range(n), key=lambda i: nums[i])

    result = [0] * n
    group_indices = [order[0]]

    def flush(group: List[int]) -> None:
        sorted_positions = sorted(group)
        sorted_values = sorted(nums[i] for i in group)
        for position, value in zip(sorted_positions, sorted_values):
            result[position] = value

    for k in range(1, n):
        previous_index = order[k - 1]
        current_index = order[k]

        if nums[current_index] - nums[previous_index] > limit:
            flush(group_indices)
            group_indices = []

        group_indices.append(current_index)

    # Process the final group.
    flush(group_indices)

    return result
````

For each adjacent critical point, calculate the difference and check if it is the minimum distance.

![2058-find-the-minimum-and-maximum-number-of-nodes-between-critical-points-2026-08-31](../assets/images/2058-find-the-minimum-and-maximum-number-of-nodes-between-critical-points-2026-08-31.png)

````python
# Definition for singly-linked list.

from typing import List, Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def build_linked_list(self, values: list):
        dummy = ListNode()
        curr = dummy
        for value in values:
            curr.next = ListNode(value)
            curr = curr.next
        return dummy.next
    
    def nodesBetweenCriticalPoints(self, head: Optional[ListNode]) -> List[int]:
        curr_node = head.next
        prev_node = head
        curr_idx = 1
        first_critical_idx = 0
        prev_critical_idx = 0
        min_distance = float('inf')
        result = [-1,-1]

        while curr_node.next is not None:
            #find if curr is critical point, local minima or maxima
            if (
                    prev_node.val<curr_node.val 
                    and curr_node.val>curr_node.next.val
                ) or (
                    prev_node.val>curr_node.val 
                    and curr_node.val<curr_node.next.val
                ):
                if prev_critical_idx == 0:
                    prev_critical_idx = curr_idx
                    first_critical_idx = curr_idx
                else:
                    min_distance = min(
                        min_distance, curr_idx - prev_critical_idx
                    )
                    prev_critical_idx = curr_idx
            prev_node = curr_node
            curr_node = curr_node.next
            curr_idx += 1

        if min_distance != float('inf'):
            max_distance = prev_critical_idx - first_critical_idx
            result = [
                min_distance,
                max_distance
            ]
        return result
````

The student starts at cell (0, 0) with 2 units of energy.

Use BFS with states (x, y, mask, e, steps), initializing with (sx, sy, 0, energy, 0), and for each move update e (–1 per step), update mask on 'L', reset e=energy on 'R', and return steps when mask == fullMask.
Hint 2
Maintain a 3D array bestEnergy[x][y][mask] storing the maximum e seen for each (x,y,mask) and skip any new state with e <= bestEnergy[x][y][mask] to prune.


````python
class Solution:
    def minMoves(self, classroom: List[str], energy: int) -> int:
        dx = [0, 1, 0, -1]
        dy = [1, 0, -1, 0]
        m = len(classroom)
        n = len(classroom[0])
        id = [[0] * n for _ in range(m)]
        sx = sy = 0
        cnt = 0
        for i in range(m):
            for j in range(n):
                if classroom[i][j] == "S":
                    sx, sy = i, j
                elif classroom[i][j] == "L":
                    id[i][j] = 1 << cnt
                    cnt += 1

        full = 1 << cnt
        bestEnergy = [
            [[-1 for _ in range(full)] for _ in range(n)] for _ in range(m)
        ]
        bestEnergy[sx][sy][0] = energy
        Info = collections.deque()
        Info.append((sx, sy, 0, energy, 0))
        while Info:
            x, y, mask, e, steps = Info.popleft()
            if mask == full - 1:
                return steps
            if e == 0:
                continue
            for d in range(4):
                nx = x + dx[d]
                ny = y + dy[d]
                if (
                    nx < 0
                    or nx >= m
                    or ny < 0
                    or ny >= n
                    or classroom[nx][ny] == "X"
                ):
                    continue
                ne = energy if classroom[nx][ny] == "R" else e - 1
                nmask = mask | id[nx][ny]
                if ne > bestEnergy[nx][ny][nmask]:
                    bestEnergy[nx][ny][nmask] = ne
                    Info.append((nx, ny, nmask, ne, steps + 1))
        return -1
````

````python
# using simulation: general case parity 
def uniformArray(nums1):
    n = len(nums1)
    def feasible(target):
        for i in range(n):
            curr_parity = nums1[i] % 2
            nxt_parity = (nums1[(i+1)%n] % 2)
            if curr_parity == target or nxt_parity == 1:
                continue
            return False
        return True
    return feasible(0) or feasible(1)
````
