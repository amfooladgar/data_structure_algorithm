You are given a 0-indexed array of integers nums.

A prefix nums[0..i] is sequential if, for all 1 <= j <= i, nums[j] = nums[j - 1] + 1. In particular, the prefix consisting only of nums[0] is sequential.

Return the smallest integer x missing from nums such that x is greater than or equal to the sum of the longest sequential prefix.

 

Example 1:

Input: nums = [1,2,3,2,5]
Output: 6
Explanation: The longest sequential prefix of nums is [1,2,3] with a sum of 6. 6 is not in the array, therefore 6 is the smallest missing integer greater than or equal to the sum of the longest sequential prefix.
Example 2:

Input: nums = [3,4,5,1,12,14,13]
Output: 15
Explanation: The longest sequential prefix of nums is [3,4,5] with a sum of 12. 12, 13, and 14 belong to the array while 15 does not. Therefore 15 is the smallest missing integer greater than or equal to the sum of the longest sequential prefix.
 

Constraints:

1 <= nums.length <= 50
1 <= nums[i] <= 50

Hint 1
To find the longest sequential prefix, iterate from left to right. For a fixed i, if nums[i] != nums[i - 1] + 1 then the longest sequential prefix ends at i - 1.

````python
# Approach 1: Simulation + Prefix Sum: O(n) time and space
def missingInteger(nums):
    n = len(nums)
    total = nums[0]
    for i in range(1,n):
        if nums[i-1]+1==nums[i]:
            total += nums[i]
        else:
            break
    nums_set = set(nums)
    while total in nums_set:
        total +=1
    return total
````

````python
# Approach 2: Simulation + Series Sum Formula
def missingInteger(nums):
    nums_set = set(nums)
    prev = nums[0]
    prefix_len = 1
    for num in nums[1:]:
        if prev + 1 == num:
            prefix_len += 1
            prev = num
        else:
            break
    total = (nums[prefix_len - 1] + nums[0]) * prefix_len // 2
    while total in nums_set:
        total+=1
    return total
````

Hint 2

````python
# Approach 1: Counting and Sliding Window: O(n) time and O(n) space (worst case)
from collections import Counter
def maxSubarrayLength(nums, k):
    n = len(nums)
    freq = Counter()

    ans, start = 0, 0
    for end in range(n):
        freq[nums[end]] += 1
        while freq[nums[end]] > k:
            freq[nums[start]] -= 1
            start+=1
        ans = max(ans, end - start+1)
    return ans
````

````python
# Approach 2: Counting and Sliding Window without Nested Loops

from collections import Counter
def maxSubarrayLength(nums, k):
    n = len(nums)
    freq = Counter()
    start = 0
    freq_more_k = 0
    for end in range(n):
        freq[nums[end]]+=1
        if freq[nums[end]] == k + 1:
            freq_more_k += 1
        if freq_more_k:
            freq[nums[start]] -= 1
            if freq[nums[start]] == k: # this guarantees that the size of our current window is equal to the largest "good" subarray encountered so far.
                freq_more_k -=1
            start += 1
    return n - start 
````

# Intuition

We use a sliding window to find the longest subarray where each number appears at most `k` times.

The window grows when it is valid. When a number appears more than `k` times, we move `start` forward once to prevent the window size from increasing.

Unlike the standard sliding-window solution, we use `if` instead of `while`. Therefore, the current window can remain temporarily invalid. This is safe because an invalid window never grows:

* Valid iteration: only `end` moves, so the window grows.
* Invalid iteration: both `end` and `start` move, so the window size stays unchanged.

Thus, the final window size `n - start` is the maximum valid length encountered.

# Approach

1. Use `freq` to count each number in the current window.

2. Use `violations` to count how many values occur more than `k` times.

3. For every `nums[end]`:

   * Add it to the window.
   * If its frequency becomes exactly `k + 1`, add one violation.
   * If a violation exists, remove `nums[start]` and move `start` once.
   * If the removed number’s frequency becomes `k`, that violation is resolved.

4. Return `n - start`, which represents the largest valid window size encountered.

```python
from collections import Counter


class Solution:
    def maxSubarrayLength(self, nums: List[int], k: int) -> int:
        n = len(nums)
        freq = Counter()
        start = 0
        violations = 0

        for end in range(n):
            freq[nums[end]] += 1

            # nums[end] has just exceeded the allowed frequency.
            if freq[nums[end]] == k + 1:
                violations += 1

            # Move start once so an invalid window cannot grow.
            if violations:
                freq[nums[start]] -= 1

                # This number is no longer violating the limit.
                if freq[nums[start]] == k:
                    violations -= 1

                start += 1

        return n - start
```

# Complexity

* Time complexity: (O(n))

  Each element is added once and removed at most once.

* Space complexity: (O(u))

  Here, (u) is the number of unique values in `nums`.

# Step-by-step solution derivation

## 1. Why a simple scan is too slow

After each update, we could scan the full string and find the longest repeating run.

* One scan takes (O(n)).
* There can be (10^5) updates.
* Total time can reach (O(nk)), or approximately (10^{10}) operations.

We need to update one position without recalculating the full string.

## 2. Use a segment tree

A segment tree divides the string into smaller intervals.

After changing one character:

* Update its leaf.
* Recalculate only its ancestors.
* Read the answer from the root.

This gives (O(\log n)) time for each query.

## 3. What must each node remember?

For its substring, each node stores:

| Field        | Meaning                                     |
| ------------ | ------------------------------------------- |
| `length`     | Number of characters in the segment         |
| `left_char`  | First character                             |
| `right_char` | Last character                              |
| `prefix`     | Longest uniform run starting from the left  |
| `suffix`     | Longest uniform run ending at the right     |
| `best`       | Longest uniform run anywhere in the segment |

For a leaf containing one character:

```text
length = prefix = suffix = best = 1
left_char = right_char = character
```

## 4. Merge two children

Suppose an interval has children `L` and `R`.

The combined run can cross their boundary only when:

```python
L.right_char == R.left_char
```

Start with values that do not cross the boundary:

```python
prefix = L.prefix
suffix = R.suffix
best = max(L.best, R.best)
```

If the boundary characters match, the crossing run is:

```python
L.suffix + R.prefix
```

Therefore:

```python
best = max(best, L.suffix + R.prefix)
```

The prefix can extend into `R` only when the complete left segment is uniform:

```python
if L.prefix == L.length:
    prefix = L.length + R.prefix
```

Similarly, the suffix can extend into `L` only when the complete right segment is uniform:

```python
if R.suffix == R.length:
    suffix = R.length + L.suffix
```

## 5. Process an update

For every query:

1. Change the corresponding leaf.
2. Recalculate the nodes on the path to the root.
3. Append `tree[1].best` to the result.

The root represents the complete string, so its `best` value is the required answer.

---

# Intuition

A character update can affect only the repeating runs that pass through that position. A segment tree lets us update that position and recompute only the affected intervals.

For each interval, we remember its best internal run and the uniform runs touching its two boundaries. When two intervals are merged, their boundary runs can connect only when the touching characters are equal.

# Approach

Build a segment tree over the characters of `s`.

Each node stores:

* The segment length.
* Its leftmost and rightmost characters.
* The longest uniform prefix.
* The longest uniform suffix.
* The longest uniform substring anywhere in the segment.

To merge nodes `L` and `R`:

1. Copy the prefix from `L` and the suffix from `R`.
2. Set the initial best result to `max(L.best, R.best)`.
3. If `L.right_char == R.left_char`:

   * Consider `L.suffix + R.prefix` as a crossing run.
   * Extend the prefix if all of `L` is uniform.
   * Extend the suffix if all of `R` is uniform.

For each query, update one leaf and recompute its ancestors. The root node’s `best` value is the answer after that query.

# Complexity

* Time complexity:

  Building the tree takes (O(n)).

  Each point update takes (O(\log n)), so all queries take:

  [
  O(n + k\log n)
  ]

* Space complexity:

  [
  O(n)
  ]

  The segment tree contains (O(n)) nodes.


Memory rule: **Two runs connect only when their boundary characters match. The prefix or suffix extends further only when the corresponding child is completely uniform.**


````python
from typing import List


class Node:
    __slots__ = (
        "length",
        "left_char",
        "right_char",
        "prefix",
        "suffix",
        "best",
    )

    def __init__(
        self,
        length=0,
        left_char="",
        right_char="",
        prefix=0,
        suffix=0,
        best=0,
    ):
        self.length = length
        self.left_char = left_char
        self.right_char = right_char
        self.prefix = prefix
        self.suffix = suffix
        self.best = best


class Solution:
    def longestRepeating(
        self,
        s: str,
        queryCharacters: str,
        queryIndices: List[int],
    ) -> List[int]:

        n = len(s)
        characters = list(s)
        tree = [Node() for _ in range(4 * n)]

        def make_leaf(char: str) -> Node:
            return Node(
                length=1,
                left_char=char,
                right_char=char,
                prefix=1,
                suffix=1,
                best=1,
            )

        def merge(left: Node, right: Node) -> Node:
            node = Node()

            node.length = left.length + right.length
            node.left_char = left.left_char
            node.right_char = right.right_char

            # Runs that do not cross the boundary.
            node.prefix = left.prefix
            node.suffix = right.suffix
            node.best = max(left.best, right.best)

            # Boundary runs can connect only when the characters match.
            if left.right_char == right.left_char:
                crossing_run = left.suffix + right.prefix
                node.best = max(node.best, crossing_run)

                # The prefix reaches the right child only when
                # the complete left child is uniform.
                if left.prefix == left.length:
                    node.prefix = left.length + right.prefix

                # The suffix reaches the left child only when
                # the complete right child is uniform.
                if right.suffix == right.length:
                    node.suffix = right.length + left.suffix

            return node

        def build(tree_index: int, segment_left: int, segment_right: int) -> None:
            if segment_left == segment_right:
                tree[tree_index] = make_leaf(characters[segment_left])
                return

            middle = (segment_left + segment_right) // 2

            build(tree_index * 2, segment_left, middle)
            build(tree_index * 2 + 1, middle + 1, segment_right)

            tree[tree_index] = merge(
                tree[tree_index * 2],
                tree[tree_index * 2 + 1],
            )

        def update(
            tree_index: int,
            segment_left: int,
            segment_right: int,
            position: int,
            new_character: str,
        ) -> None:
            if segment_left == segment_right:
                tree[tree_index] = make_leaf(new_character)
                return

            middle = (segment_left + segment_right) // 2

            if position <= middle:
                update(
                    tree_index * 2,
                    segment_left,
                    middle,
                    position,
                    new_character,
                )
            else:
                update(
                    tree_index * 2 + 1,
                    middle + 1,
                    segment_right,
                    position,
                    new_character,
                )

            tree[tree_index] = merge(
                tree[tree_index * 2],
                tree[tree_index * 2 + 1],
            )

        build(1, 0, n - 1)

        answers = []

        for position, new_character in zip(
            queryIndices,
            queryCharacters,
        ):
            characters[position] = new_character
            update(1, 0, n - 1, position, new_character)
            answers.append(tree[1].best)

        return answers
````

````python
# Leetcode solution:
class Solution:
    def longestRepeating(
        self, s: str, queryCharacters: str, queryIndices: List[int]
    ) -> List[int]:
        n = len(s)
        pre = [0] * (4 * n)
        suf = [0] * (4 * n)
        maxLen = [0] * (4 * n)
        leftChar = [""] * (4 * n)
        rightChar = [""] * (4 * n)

        def build(u: int, l: int, r: int) -> None:
            if l == r:
                pre[u] = 1
                suf[u] = 1
                maxLen[u] = 1
                leftChar[u] = s[l]
                rightChar[u] = s[l]
                return
            mid = (l + r) >> 1
            build(u << 1, l, mid)
            build(u << 1 | 1, mid + 1, r)
            pushUp(u, l, r)

        def pushUp(u: int, l: int, r: int) -> None:
            mid = (l + r) >> 1
            leftLen = mid - l + 1
            rightLen = r - mid
            left = u << 1
            right = u << 1 | 1
            leftChar[u] = leftChar[left]
            rightChar[u] = rightChar[right]
            pre[u] = pre[left]
            if pre[left] == leftLen and rightChar[left] == leftChar[right]:
                pre[u] = pre[left] + pre[right]
            suf[u] = suf[right]
            if suf[right] == rightLen and rightChar[left] == leftChar[right]:
                suf[u] = suf[right] + suf[left]
            maxLen[u] = max(maxLen[left], maxLen[right])
            if rightChar[left] == leftChar[right]:
                maxLen[u] = max(maxLen[u], suf[left] + pre[right])

        def update(u: int, l: int, r: int, pos: int, ch: str) -> None:
            if l == r:
                leftChar[u] = ch
                rightChar[u] = ch
                return
            mid = (l + r) >> 1
            if pos <= mid:
                update(u << 1, l, mid, pos, ch)
            else:
                update(u << 1 | 1, mid + 1, r, pos, ch)
            pushUp(u, l, r)

        build(1, 0, n - 1)
        k = len(queryIndices)
        ans = []
        for i in range(k):
            update(1, 0, n - 1, queryIndices[i], queryCharacters[i])
            ans.append(maxLen[1])
        return ans
````

````python
# Brute force/enumeration: O(n^2) time and O(1) space
def maximumLengthSubstring(s):
    n = len(s)
    ans = 0
    for l in range(n):
        count = [0]*26
        for r in range(l, n):
            ch = ord(s[r]) - ord('a')
            count[ch] += 1
            if count[ch]>2:
                break
            ans = max(ans, r - l + 1)
    return ans
````

````python
from collections import defaultdict

def maximumLengthSubstring(s: str) -> int:
    count = defaultdict(int)
    left = 0
    ans = 0

    for right, char in enumerate(s):
        count[char] += 1

        # Shrink until the current character appears at most twice.
        while count[char] > 2:
            count[s[left]] -= 1
            left += 1

        ans = max(ans, right - left + 1)

    return ans
````

````python
# Approach: Classification Discussion
def longestSubsequence(nums):
    totalXor = 0
    n = len(nums)
    allZero = True
    for num in nums:
        totalXor ^= num
        if num > 0:
            allZero = False

    if totalXor > 0:
        return n
    return n - 1 if allZero == False else 0
````

Alice and Bob continue their games with stones. There is a row of n stones, and each stone has an associated value. You are given an integer array stones, where stones[i] is the value of the ith stone.

Alice and Bob take turns, with Alice starting first. On each turn, the player may remove any stone from stones. The player who removes a stone loses if the sum of the values of all removed stones is divisible by 3. Bob will win automatically if there are no remaining stones (even if it is Alice's turn).

Assuming both players play optimally, return true if Alice wins and false if Bob wins.

 

Example 1:

Input: stones = [2,1]
Output: true
Explanation: The game will be played as follows:
- Turn 1: Alice can remove either stone.
- Turn 2: Bob removes the remaining stone. 
The sum of the removed stones is 1 + 2 = 3 and is divisible by 3. Therefore, Bob loses and Alice wins the game.
Example 2:

Input: stones = [2]
Output: false

Since all the stones are removed and the sum of values is not divisible by 3, Bob wins the game.
Example 3:

Input: stones = [5,1,2,4,3]
Output: false
Explanation: Bob will always win. One possible way for Bob to win is shown below:
- Turn 1: Alice can remove the second stone with value 1. Sum of removed stones = 1.
- Turn 2: Bob removes the fifth stone with value 3. Sum of removed stones = 1 + 3 = 4.
- Turn 3: Alices removes the fourth stone with value 4. Sum of removed stones = 1 + 3 + 4 = 8.
- Turn 4: Bob removes the third stone with value 2. Sum of removed stones = 1 + 3 + 4 + 2 = 10.
- Turn 5: Alice removes the first stone with value 5. Sum of removed stones = 1 + 3 + 4 + 2 + 5 = 15.
Alice loses the game because the sum of the removed stones (15) is divisible by 3. Bob wins the game.
 

Constraints:

1 <= stones.length <= 105
1 <= stones[i] <= 104

Hint 1
There are limited outcomes given the current sum and the stones remaining.
Hint 2
Can we greedily simulate starting with taking a stone with remainder 1 or 2 divided by 3?

````python
from typing import List


class Solution:
    def stoneGameIX(self, stones: List[int]) -> bool:
        count = [0, 0, 0]

        for stone in stones:
            count[stone % 3] += 1

        c0, c1, c2 = count

        if c0 % 2 == 0:
            return c1 > 0 and c2 > 0

        return abs(c1 - c2) > 2
````

Hint 3

Hint 4
If 1 < k < n, all elements different from nums[0] and nums[n - 1] will occur in more than one subarray of size k. Hence, the answer is the largest of nums[0] and nums[n - 1] if they both occur exactly once in the array. If one of them occurs more than once, return the other. If both of them occur more than once, return -1.

### Straight solution
![3471-find-the-largest-almost-missing-integer-2026-08-18.png](../assets/images/3471-find-the-largest-almost-missing-integer-2026-08-18.png)

````python
# sliding window: complexity O(nk) time and O(n) space
from collections import defaultdict
def largestInteger(nums, k):
    n = len(nums)
    seen = defaultdict(int)

    for i in range(n - k + 1):
        # Count each number at most once per subarray
        unique_nums = set(nums[i:i + k])

        for num in unique_nums:
            seen[num] += 1

    ans = -1

    for num, count in seen.items():
        if count == 1:
            ans = max(ans, num)

    return ans
````

````python
# You can reduce the solution from O(nk) to O(n) by using how many windows can contain each position. 
# For 1 < k < n, only nums[0] and nums[-1] can possibly appear in exactly one window. 
# Every interior position belongs to at least two size-k windows.
# Complexity: O(n) time and space
from collections import Counter
def largestInteger(nums, k):
    n = len(nums)

    # There is only one subarray of size n.
    # Every number appears in exactly that one subarray.
    if k == n:
        return max(nums)

    freq = Counter(nums)

    # Each element forms its own subarray.
    # We need the largest globally unique number.
    if k == 1:
        return max((num for num in nums if freq[num] == 1), default=-1)

    # # OR this way (more optimized, as it stops right after finding the largest, but it is using the constraint assumption)
    # if k == 1:
    #     for i in range(50, -1, -1):
    #         if count[i] == 1:
    #             return i
    #     return -1

    # For 1 < k < n:
    # only the first and last elements can belong to exactly one window.
    ans = -1

    if freq[nums[0]] == 1:
        ans = nums[0]

    if freq[nums[-1]] == 1:
        ans = max(ans, nums[-1])

    return ans
````
